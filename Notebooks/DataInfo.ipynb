## Number of book Ratings -> 0 presence

## Book author by books

```python
authors_by_books = books.groupby("Book-Author")["Book-Title"].count().sort_values(ascending=False)
authors_by_books
```

```python
print("\nTop 10 authors by number of books:")
print(authors_by_books.head(10))
```

```python
## Top five user rated
rated_users = ratings.groupby("User-ID")["Book-Rating"].count().sort_values(ascending=False)
print("\nTop 10 users by number of ratings:")
print(rated_users.head(10))
```

```python
## Totaol ratings per book
ratings_per_book = ratings.groupby("ISBN")["Book-Rating"].count().sort_values(ascending=False)
books_with_ratings = books[books["ISBN"].isin(ratings_per_book.index)]
merged_data = pd.merge(books_with_ratings, ratings_per_book.reset_index(), on="ISBN", how="inner")
print("\nTop 10 books by number of ratings:")
print(ratings_per_book.head(10))
print("\nTop 10 books with their ratings count:")
print(merged_data[["Book-Title", "Book-Rating"]].sort_values("Book-Rating", ascending=False).head(10))
```

```python
max_book_publishers = books.groupby("Publisher")["Book-Title"].count().sort_values(ascending=False)
print("\nTop 10 publishers by number of books published:")
print(max_book_publishers.head(10))
```

```python
count_books_per_year = books.groupby("Year-Of-Publication")["Book-Title"].count().sort_values(ascending=False)
print("\nTop 10 years by number of books published:")
print(count_books_per_year.head(10))
```
